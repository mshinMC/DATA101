```python
##### Part 1:

import numpy as np


# Creates a list for labeling each assessment type
assessments = ['Homework', 'Project', 'Exam 1', 'Exam 2', 'Final Exam']
# Creates an empty list to record all student names
students = []
# Creates an empty list to temporarily store scores for each assessment
scoreSet = []

# Creates a matrix of zeros as a placeholder for all students' scores;
# 10 rows for each student to store all their scores, 5 columns for each assessment type 
allGrades = np.zeros((10,5))


# Loops for each row in the "allGrades" matrix, once for each student
for x in range(0, 10):
    
    # Starts each loop by emptying "scoreSet" list to make room for each new student's scores
    scoreSet.clear()
    
    # Asks to input student's name, then adds the name to "students" list 
    students.append(input("Enter student name:"))
    
    # Loops for each assessment type
    for y in assessments:
        # Asks to input a score, then adds the value to "scoreSet" temporary list;
        scoreSet.append(input(f"Enter {y} score:"))
        # 'y' variable of the loop connects to "assessments" list; used to format-print when asking for input;
        # made to ensure that scores are entered in order to match with assessment type/weight
    # {end 'y' loop}
    
    # Copies the values from "scoreSet" list to the student's respective row in "allGrades" matrix
    allGrades[x,] = scoreSet.copy()
    print()

# {end of 'x' loop for recording student names/scores}



# Creates array for each assessment type's score weight
scoreWeights = np.array([0.2, 0.15, 0.2, 0.2, 0.25])
# Creates an array of zeros as a placeholder to hold each students' final (weighted) grade;
# 10 slots, one for each student
finalGrades = np.zeros(10)


# Loops once for each student
for a in range(0,10):
# 'a' index matches with each student's row in "allGrades" matrix
    
    # Sets a temporary variable (to store accumulated weighted scores) to value of 0.0;
    # start of each loop resets the total, so only scores for the student that comes next are stored
    weightTotal = 0.0

    # Loops for each assessment type
    for b in range(0,5):
    # 'b' index coordinates with each assessment type
        
        # Calculates weighted score for an assessment, then stores value temporarily to variable; 
        # multiply [base score] by respective [assessment weight]
        weightedScore = float(allGrades[a,b] * scoreWeights[b])
        # [base score] drawn from "allGrades" matrix;
        # indexed from inner and outer loop variables ('a' and 'b')
        # [assessment weight] drawn from "scoreWeights" array
        # indexed from inner loop variable ('b')

        # Adds weighted score from each assessment to the total
        weightTotal = float(weightTotal + weightedScore)
        # When the 'b' loop re-iterates, the variable holding the 
        # weighted score resets, so that the next assessment's score can be added to the total
    # {end 'b' loop}

    # Inputs the final weighted grade for the respective student into array of final (weighted) grades;
    # indexed from 'a' variable so the array values match up with each student
    finalGrades.put(a, weightTotal)

# {end of 'a' loop for calculating and storing student final grades}



# Creates an empty list to store each student's final grade as a letter
gradeLetters = []

# Loops for each final grade
for z in finalGrades:
    # Compares grade value in 'z' index of "finalGrades" array to each letter's percent range
    
    if 90 <= z <= 100:
        gradeLetters.append('A')
        # checks if grade is between 90%-100%; if true, then assign letter A and store in
        # "gradeLetters" list for respective student's final grade 
    
    elif 80 <= z < 90:
        gradeLetters.append('B')
        # if grade is between 80%-89%, then assign letter B and store, {etc.}
    
    elif 70 <= z < 80:
        gradeLetters.append('C')
        # if between 70-79, then letter C, {etc.}
    
    elif 60 <= z < 70:
        gradeLetters.append('D')
        # between 60-69 for letter D, {etc.}
    
    elif z < 60:
        gradeLetters.append('F')
        # if grade if less than 60, then assign letter F, {etc.}

# {end of 'z' loop for determining student final grade letters}


# Loops for each student
for c in range(0, 10):
# 'c' index matches with each student and their respective final grade
    
    # Displays each student with their [name] and [final grade letter]
    print(f"{students[c]} - Grade: {gradeLetters[c]}")
    # [name] drawn from "students" list
    # [final grade letter] drawn from "gradeLetters" list

```
