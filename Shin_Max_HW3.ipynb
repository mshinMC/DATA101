```python
import numpy as np
import pandas as pd

### Part 1

titanic = pd.read_csv('titanic.csv')
titanic.rename(columns = {'PassengerId':'P.Id', 'Pclass':'P.Class', 'SibSp':'Sib-Sp', 'Parch':'Par-Ch'}, inplace = True)
titanic.info()
print()
```

```python
### Part 2


embarkedMode = titanic['Embarked'].mode()[0]
print(f"Embarked Mode Value: {embarkedMode}")
print()

titanic.fillna({'Embarked':embarkedMode}, inplace=True)
print(titanic['Embarked'].iloc[45:65])


print()
print()


ageMean = titanic['Age'].mean()
print(f"Overall Mean Age: {ageMean}")
print()

titanic['Age'] = titanic['Age'].fillna(ageMean)
print(titanic['Age'].iloc[:25])
print()
```

```python
### Part 3


FamilySize = titanic['Sib-Sp'] + titanic['Par-Ch'] + 1
print(FamilySize)

print()

minFamilySize = FamilySize.min()
maxFamilySize = FamilySize.max()
avgFamilySize = FamilySize.mean()

print(f"Minimum Family Size: {minFamilySize}")
print(f"Maximum Family Size: {maxFamilySize}")
print(f"Average Family Size: {avgFamilySize}")

print()

# creates series from "FamilySize" containing all passengers with family size of 1
noFamily = FamilySize[FamilySize == 1]
# gets length of "noFamily", which is the number of passengers that are alone
P_Alone = len(noFamily)
print(f"Number of Passengers Traveling Alone: {P_Alone}")
print()
```

```python
### Part 4


# creates dataframe filtered by passengers less than 16 years old, and without N/A value as 'Sex'
filtered = titanic.query("Age < 16 and (Sex == 'male' or Sex == 'female') ")
print(filtered.head())

print()
print()

survivalRate = filtered['Survived'].mean()
print(f"Survival Rate: {survivalRate}")
print()


# survival rate for each combined category of 'Sex' and 'P.Class' (passenger class)
sortedSurvivalRate = filtered.groupby(['Sex', 'P.Class'])['Survived'].agg('mean')
```

```python
### Part 5


# when sorting by ticket, determines the number of people associated with each unique ticket, and grabs it's fare price  
ticketFareDivider = titanic.groupby('Ticket').agg(peoplePerTicket = ('P.Id','count'), ticketFare = ('Fare', 'median'))
# 'median' function is used with .agg() since 'Fare' value will be the same for every entry of the same ticket


# calculates fare price per passenger on a ticket by dividing ticket total fare by number of people on the ticket
ticketFareDivider['pricePerPassenger'] = ticketFareDivider['ticketFare'] / ticketFareDivider['peoplePerTicket']


# appends each ticket's passenger class
ticketFareDivider['ticketClass'] = titanic.groupby('Ticket')['P.Class'].agg('median')
# 'median' function is used to aggregate for same reason as 'ticketFare' column

print(ticketFareDivider)

print()
print()

avgClassFare = ticketFareDivider.groupby('ticketClass')['pricePerPassenger'].agg('mean')
print(avgClassFare)
print()
```
