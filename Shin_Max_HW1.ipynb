```python
# creates a string variable for the product's name
productName = "peanut butter"
# creates a float variable for the product's price
productPrice = 3.75
# creates an integer variable for the product's quantity in stock
productStock = 20

# changes the product's price into an integer value
productPrice = int(productPrice)
# prints the data type of the variable for the price
print(type(productPrice))

# prints out a sentence displaying the product, it's price, and how much is in stock
print(f"We have {productStock} units of {productName} available at ${productPrice} each.")
```

```python
# calculates total value of the inventory for the product by multiplying it's price by the stock
totalValue = productPrice * productStock
# prints the total value of the inventory
print(totalValue)

# checks to see if the total value is greater than 500, then prints the result (True or False)
print(totalValue > 500)
```
